```python
students = [
    {"name": "John", "age": 16, "gender": "Male", "score": 75},
    {"name": "Mary", "age": 15, "gender": "Female", "score": 62},
    {"name": "David", "age": 17, "gender": "Male", "score": 48},
    {"name": "Grace", "age": 16, "gender": "Female", "score": 85},
    {"name": "Peter", "age": 20, "gender": "Male", "score": 50}
]

for student in students:
    score = student["score"]

    if score >= 70:
        grade = "Excellent"
    elif score >= 50:
        grade = "Pass"
    else:
        grade = "Fail"

    print("Name:", student["name"])
    print("Age:", student["age"])
    print("Gender:", student["gender"])
    print("Score:", score)
    print("grade:", grade)
    print("------------------")
```
