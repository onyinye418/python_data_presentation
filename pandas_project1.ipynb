```python
# ASSIGNMENT
df1 = pd.read_csv("c:/Users/HomePC/Desktop/PYTHON DATA ANALYSIS/Datasets/product_dataset.csv")
```

```python
print (df1)
```

```python
# Creating New Columns
df1["Total"] = df1["Quantity"] * df1["Price"]
print (df1)
```

```python
# Applying Functions
df1["Total_Percentage"] = df1["Total"].apply(lambda x: (x / 100) * 100)
print(df1)
```

```python
df1["Score_Percentage"] = (df1["Total"] / 233592.00) * (100/1)
# print(df1["Score_Percentage"])
print(df1)
```

```python
df1.columns
```

```python
df1["Score_Percentage"]
```

```python
# average price by category
categorysum=df1.groupby("Category")["Total"].mean() 
print(categorysum)
```

```python
category_totals = df1.groupby("Category")["Total"].sum()

category_totals.plot(kind="bar", xlabel="Category", ylabel="Total Score")
```

```python
df1["Category"]
```

```python
# average price by product name
product_total = df1.groupby("Product_Name")["Total"].sum() 
product_total.plot(kind="bar", xlabel="Product_Name", ylabel="Total Amount", title="Revenue by Product" )
```

```python
# Multiple aggregations at once
df1.groupby("Product_Name")["Total"].agg(["mean", "min", "max", "count"])
```

```python
# Multiple aggregations at once
aggregate = df1.groupby("Category")["Total"].agg(["mean", "min", "max", "count"])

print(aggregate)
```

```python
import matplotlib.pyplot as plt
```

```python
summary = df1.groupby("Category")["Total"].mean()

plt.bar(summary.index, summary.values)
plt.xlabel("Category")
plt.ylabel("Mean Total")
plt.title("Mean Total by Category")
figsize=(2, 2),
plt.show()
```

```python
summary = df1.groupby("Category")["Total"].agg(["mean", "min", "max", "count"])

summary.plot(
    kind="bar",
    subplots=True,
    layout=(2, 2),
    figsize=(5, 5),
    legend=False
)

plt.tight_layout()
plt.show()
```

```python
department_summary = (
    df1.groupby("Category")["Total"]
       .sum()  # or .mean(), depending on what you want
       .sort_values(ascending=False)
)

print(department_summary)
```

```python
df1.columns
```

```python
Product_summary = (
    df1.groupby("Product_Name")["Total"]
       .sum()  # or .mean(), depending on what you want
       .sort_values(ascending=False)
)

print(Product_summary)
```

```python
Product_summary.sort_values().plot(kind="barh", figsize=(5, 4))

plt.xlabel("Total")
plt.ylabel("Product")
plt.title("Total sales by product")
plt.tight_layout()
plt.show()
```

```python
# highest sales by supplier
highest_total = df1["Supplier"].max(),
lowest_total = df1["Supplier"].min()
print(highest_total)
print(lowest_total)
```

```python
supplier_totals = df1.groupby("Supplier")["Total"].sum()

highest_supplier = supplier_totals.idxmax()
lowest_supplier = supplier_totals.idxmin()

selected = supplier_totals.loc[[highest_supplier, lowest_supplier]]

selected.plot(kind="bar", color=["green", "red"], figsize=(5, 3))

plt.xlabel("Supplier")
plt.ylabel("Total")
plt.title("Highest and Lowest Total by Supplier")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
# Sales by suppliers
sales_by_supplier = df1.groupby("Supplier")["Total"].sum()
print(sales_by_supplier)
```

```python
sales_by_supplier.plot(kind="bar", xlabel="Supplier", ylabel="Supplier Revenue", title="Revenue by Supplier")
```
