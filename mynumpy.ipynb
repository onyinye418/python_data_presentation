```python
import numpy as np

np.random.seed(10)  
scores = np.random.randint(0, 101, 100)  

print(scores)
```

```python
print("Average score:", scores.mean())

print("Highest score:", scores.max())

print("Lowest score:", scores.min())

print("Students above 70:", np.sum(scores > 70))

print("Students who failed (< 50):", np.sum(scores < 50))

print("students who pass (> 50):", np.sum(scores > 50))
```
